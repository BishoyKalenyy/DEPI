```python
%pip install -q langchain langgraph langchain-groq pydantic python-dotenv

import os
import random
import json
from datetime import date, timedelta
from typing import Literal

from dotenv import load_dotenv
load_dotenv()  # reads GROQ_API_KEY from .env

if not os.environ.get("GROQ_API_KEY"):
    raise RuntimeError("Set GROQ_API_KEY in your .env file")

MODEL = "groq:openai/gpt-oss-20b"  # swap for a cheaper/bigger Groq model if you like

from langchain.agents import create_agent, AgentState
from langchain.tools import tool
from pydantic import BaseModel, Field

print("Ready. Model:", MODEL)
```

```python
@tool
def get_weather(city: str) -> str:
    """Return the (fake) current weather for a city."""
    return f"It's sunny and 27°C in {city}."


test_agent = create_agent(
    model=MODEL,
    tools=[get_weather],
    system_prompt="You are a terse, friendly assistant. Use tools when they help.",
)

result = test_agent.invoke(
    {"messages": [{"role": "user", "content": "What's the weather in Cairo?"}]}
)
print(result["messages"][-1].content)
```
